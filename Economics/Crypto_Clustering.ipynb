```python
# libraries
import pandas as pd
import hvplot.pandas
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
df_market_data = pd.read_csv(
    r"C:\Users\secar\OneDrive\Documents\Bootcamp\19-Unsupervised-Learning\Resources\crypto_market_data.csv",
    index_col="coin_id"
)

# Display sample data
df_market_data.head(10)
```

```python
# summary statistics
df_market_data.describe()
```

```python
# Plot 
df_market_data.hvplot.line(
    width=800,
    height=400,
    rot=90
)
```

---

### Prepare the Data

```python
# `StandardScaler()` module from scikit-learn to normalize the data from the CSV file
market_data_scaled = StandardScaler().fit_transform(df_market_data)
```

```python
# DataFrame with the scaled data
# Set the index to the original `coin_id` values.
df_market_data_scaled = pd.DataFrame(
    market_data_scaled,
    columns=df_market_data.columns,
    index=df_market_data.index
)

# sample data
df_market_data_scaled.head()
```

```python
print(df_market_data.columns)
```

---

### Find the Best Value for k Using the Original Data.

```python
# list with the number of k-values from 1 to 11
k_values = list(range(1, 12))
```

```python
# empty list to store the inertia values
inertia = []

# for loop to compute the inertia with each possible value of k
for k in k_values:
    # Create a KMeans model using the loop counter for n_clusters
    model = KMeans(n_clusters=k, random_state=42, n_init="auto")
    
    # Fit the model to the scaled data
    model.fit(df_market_data_scaled)
    
    # Append the model.inertia_ to the inertia list
    inertia.append(model.inertia_)
```

```python
# dictionary with the data to plot the Elbow curve
elbow_data = {"k": k_values, "inertia": inertia}

# DataFrame with the data to plot the Elbow curve
elbow_df = pd.DataFrame(elbow_data)

elbow_df
```

```python
# line chart with all the inertia values computed with 
# the different values of k to visually identify the optimal value for k.
elbow_df.hvplot.line(
    x="k",
    y="inertia",
    title="Elbow Curve - Original Scaled Data",
    xticks=k_values
)
```

#### Answer the following question: 

**Question:** What is the best value for `k`?

**Answer:** 4

---

### Cluster Cryptocurrencies with K-means Using the Original Data

```python
# Initialize the K-Means model using the best value for k
best_k = 4

kmeans_model = KMeans(n_clusters=best_k, random_state=42)
```

```python
# Fit the K-Means model using the scaled data
kmeans_model.fit(df_market_data_scaled)
```

```python
# Predict the clusters to group the cryptocurrencies using the scaled data
market_data_clusters = kmeans_model.predict(df_market_data_scaled)

# Print the resulting array of cluster values.
market_data_clusters
```

```python
# Copy of the original DataFrame
df_market_data_predictions = df_market_data.copy()
```

```python
# Add a new column to the DataFrame with the predicted clusters
df_market_data_predictions["predicted_cluster"] = market_data_clusters

# Sample data
df_market_data_predictions.head()
```

```python
# Create a scatter plot using hvPlot by setting 
# `x="price_change_percentage_24h"` and `y="price_change_percentage_7d"`. 
# Color the graph points with the labels found using K-Means and 
# add the crypto name in the `hover_cols` parameter to identify 
# the cryptocurrency represented by each data point.
df_market_data_predictions.reset_index().hvplot.scatter(
    x="price_change_percentage_24h",
    y="price_change_percentage_7d",
    by="predicted_cluster",
    hover_cols=["coin_id"],
    title="Cryptocurrency Clusters - Original Scaled Data"
)
```

### Optimize Clusters with Principal Component Analysis.

```python
# Create a PCA model instance and set `n_components=3`.
pca_model = PCA(n_components=3)
```

```python
# Use the PCA model with `fit_transform` to reduce to 
# three principal components.
market_data_pca = pca_model.fit_transform(df_market_data_scaled)

# View the first five rows of the array. 
market_data_pca[:5]
```

```python
# Retrieve the explained variance to determine how much information 
# can be attributed to each principal component.
pca_model.explained_variance_ratio_
```

#### Answer the following question: 

**Question:** What is the total explained variance of the three principal components?

**Answer:** The total explained variance of the three principal components is the sum of these variances, which is 0.8950. 
You can arrive at the same answer by using the explained variance ratio

```python
#The total explained variance of the three principal components is the sum of these variances, which is 0.8950. 
#You can arrive at the same answer by using the explained variance ratio
```

```python
# Cew DataFrame with the PCA data.
df_market_data_pca = pd.DataFrame(
    market_data_pca,
    columns=["PC1", "PC2", "PC3"],
    index=df_market_data.index
)

# Display sample data
df_market_data_pca.head()
```

---

### Find the Best Value for k Using the PCA Data

```python
# Create a list with the number of k-values from 1 to 11
k_values = list(range(1, 12))
```

```python
# Create an empty list to store the inertia values
pca_inertia = []

# Create a for loop to compute the inertia with each possible value of k
for k in k_values:
    # Create a KMeans model using the loop counter for n_clusters
    model = KMeans(n_clusters=k, random_state=42, n_init="auto")
    
    # Fit the model to the PCA data
    model.fit(df_market_data_pca)
    
    # Append the model.inertia_ to the inertia list
    pca_inertia.append(model.inertia_)
```

```python
# Create a dictionary with the data to plot the Elbow curve
pca_elbow_data = {"k": k_values, "inertia": pca_inertia}

# Create a DataFrame with the data to plot the Elbow curve
pca_elbow_df = pd.DataFrame(pca_elbow_data)

pca_elbow_df
```

```python
# Plot a line chart with all the inertia values computed with 
# the different values of k to visually identify the optimal value for k.
pca_elbow_df.hvplot.line(
    x="k",
    y="inertia",
    title="Elbow Curve - PCA Data",
    xticks=k_values
)
```

#### Answer the following questions: 

* **Question:** What is the best value for `k` when using the PCA data?

  * **Answer:** 4


* **Question:** Does it differ from the best k value found using the original data?

  * **Answer:** No

### Cluster Cryptocurrencies with K-means Using the PCA Data

```python
# Initialize the K-Means model using the best value for k
best_k = 4

kmeans_nextmodel = KMeans(n_clusters=best_k, random_state=42)
```

```python
# Fit the K-Means model using the PCA data
kmeans_nextmodel.fit(df_market_data_pca)
```

```python
# Predict the clusters to group the cryptocurrencies using the PCA data
pca_clusters = kmeans_nextmodel.predict(df_market_data_pca)

# Print the resulting array of cluster values.
pca_clusters
```

```python
# Create a copy of the DataFrame with the PCA data
df_market_data_pca_predictions = df_market_data_pca.copy()

# Add a new column to the DataFrame with the predicted clusters
df_market_data_pca_predictions["predicted_cluster"] = pca_clusters

# Display sample data
df_market_data_pca_predictions.head()
```

```python
# Create a scatter plot using hvPlot by setting 
# `x="PC1"` and `y="PC2"`. 
# Color the graph points with the labels found using K-Means and 
# add the crypto name in the `hover_cols` parameter to identify 
# the cryptocurrency represented by each data point.
df_market_data_pca_predictions.reset_index().hvplot.scatter(
    x="PC1",
    y="PC2",
    by="predicted_cluster",
    hover_cols=["coin_id"],
    title="Cryptocurrency Clusters - PCA Data"
)
```

### Visualize and Compare the Results

In this section, you will visually analyze the cluster analysis results by contrasting the outcome with and without using the optimization techniques.

```python
# Composite plot to contrast the Elbow curves
elbow_original_plot = elbow_df.hvplot.line(
    x="k",
    y="inertia",
    title="Elbow Curve - Original Scaled Data",
    xticks=k_values
)

elbow_pca_plot = pca_elbow_df.hvplot.line(
    x="k",
    y="inertia",
    title="Elbow Curve - PCA Data",
    xticks=k_values
)

elbow_original_plot + elbow_pca_plot
```

```python
# Composite plot to contrast the clusters
clusters_original_plot = df_market_data_predictions.reset_index().hvplot.scatter(
    x="price_change_percentage_24h",
    y="price_change_percentage_7d",
    by="predicted_cluster",
    hover_cols=["coin_id"],
    title="Clusters - Original Scaled Data"
)

clusters_pca_plot = df_market_data_pca_predictions.reset_index().hvplot.scatter(
    x="PC1",
    y="PC2",
    by="predicted_cluster",
    hover_cols=["coin_id"],
    title="Clusters - PCA Data"
)

clusters_original_plot + clusters_pca_plot
```

#### Answer the following question: 

  * **Question:** After visually analyzing the cluster analysis results, what is the impact of using fewer features to cluster the data using K-Means?

  * **Answer:** Using PCA keeps most of the information from the original features while making the clusters easier to visualize and interpret. In this case, the three principal components explain about 89.5% of the total variance, and the PCA clustering produces a very similar cluster structure to the original scaled data. The main benefit is simplification: fewer dimensions, cleaner plots, and less noise, with little loss of clustering signal.
